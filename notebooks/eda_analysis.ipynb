# Day 3 (Part A) — Exploratory Data Analysis (EDA)
Real Estate Price Prediction (Gujrat, Gujranwala, Sialkot — Zameen.com)

Is notebook mein hum `clean_data.csv` ko visually explore karte hain taake data ka
pattern samajh sakein — ye insights README mein bhi kaam aayenge.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

df = pd.read_csv("../data/clean_data.csv")
df.head()
```

## 1. Price distribution

```
plt.figure(figsize=(8, 5))
sns.histplot(df["Price_PKR"] / 1e7, bins=40, kde=True)
plt.xlabel("Price (Crore PKR)")
plt.title("Price Distribution")
plt.show()
```

## 2. Price vs Size

```
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Size_Marla", y=df["Price_PKR"] / 1e7, hue="Property_Type", alpha=0.6)
plt.xlabel("Size (Marla)")
plt.ylabel("Price (Crore PKR)")
plt.title("Price vs Size")
plt.show()
```

## 3. Average price by city

```
avg_price_by_city = df.groupby("City", observed=True)["Price_PKR"].mean().sort_values() / 1e7

plt.figure(figsize=(7, 4))
avg_price_by_city.plot(kind="barh")
plt.xlabel("Average Price (Crore PKR)")
plt.title("Average Price by City")
plt.show()
```

## 4. Average price per Marla by city
Ye location ki 'per-unit value' behtar dikhata hai, jitni bari property utna hi zyada total price expected hota hai isliye ye comparison zyada fair hai.

```
avg_ppm_by_city = df.groupby("City", observed=True)["Price_per_Marla"].mean().sort_values() / 1e5

plt.figure(figsize=(7, 4))
avg_ppm_by_city.plot(kind="barh", color="orange")
plt.xlabel("Average Price per Marla (Lakh PKR)")
plt.title("Average Price per Marla by City")
plt.show()
```

## 5. Property type breakdown

```
plt.figure(figsize=(6, 6))
df["Property_Type"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.ylabel("")
plt.title("Property Type Breakdown")
plt.show()
```

## 6. Correlation heatmap
Sirf numeric columns pe — dekhte hain kaunse features price se sabse zyada related hain.

```
numeric_cols = ["Price_PKR", "Size_Marla", "Price_per_Marla"]
corr = df[numeric_cols].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

Har graph dekhne ke baad, apni observations yahan likh lein, jaise:
